## Inconsistent Data

Inconsistent data is likely due to inconsistent formatting and can be addressed by re-formatting all values in a column or row.

```python
# Look out for inconsistent data!
df["Division Name"].unique()


```

```python
df["Department Name"].unique()
```

```python
df["Class Name"].unique()
```

```python
df["Class Name"].value_counts()
```

Did you find any inconsistent data? What did you do to clean it?

```python
# Make your notes here!
# Yes, found a typo in one of the categories, there are very similar and redundant class names, including two that only contain 1 and 2 values (casual botoms). 
```

```python
df["Division Name"] = df["Division Name"].replace("Initmates", "Intimates") # fixed typo in "intimates"
```

```python
df["Division Name"] = df["Division Name"].str.title()
df["Department Name"] = df["Department Name"].str.title()
df["Class Name"] = df["Class Name"].str.title()

# standardized capitalization formatting
```

```python
df["Class Name"] = df["Class Name"].replace({"Casual Bottoms": "Pants", "Chemises": "Sleep"})

# merged unecessary catagories. 
```
